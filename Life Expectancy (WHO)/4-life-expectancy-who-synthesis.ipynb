## 4.2-Statistical Confirmation of Structural Health Disparities

The analyses conducted in Notebook 3 provide the **statistical foundation** for interpreting the descriptive results in Notebook 4. Notebook 3 examined population health differences using both **cross-sectional data for 2015** and **panel data from 2000 to 2015**. Key points supporting subsequent analyses are summarized below.

---

**Significant Differences by Development Status**
- Developed vs. developing countries show a **large, statistically significant separation** in the composite health index (PC1_health_index) in 2015 (**U = 479, p < 0.001; δ = −0.82**).  
- Developed countries consistently display **lower PC1 values**, corresponding to better overall population health.  
- Although robust regression adjusting for Schooling was applied, the high negative Spearman correlation between PC1_health_index and Schooling (**ρ ≈ −0.74**) limits interpretability. *The adjusted estimates are influenced by multicollinearity, so differences cannot be fully attributed to education alone.*  
- The model mainly demonstrates the **methodological approach under non-normality and outliers**, rather than a causal decomposition.

---

**Regional Disparities Are Statistically Robust**
- Kruskal–Wallis tests reveal highly significant differences in health outcomes across regions (**H = 108.09, p < 0.001**).  
- Post-hoc Dunn’s tests identify which regions differ significantly:  
  - **Africa and the Middle East** systematically have worse outcomes  
  - **Europe, North America, and Oceania** have relatively better and similar outcomes  
- These findings justify **visual comparisons across regions** and identification of outlier countries.

---

**Temporal and Panel Analyses Confirm Structural Patterns**
- From 2000 to 2015, regional trajectories of PC1_health_index remain relatively stable.  
- **Between-region variance (≈ 1.04)** dominates **within-region fluctuations over time (≈ 0.067)**.  
- Normalized trends show that regional rankings are persistent, indicating that **2015 differences reflect enduring structural disparities**.  
- Panel regressions with fixed effects further demonstrate **robust associations between education (Schooling) and regional health**, controlling for unobserved time-invariant characteristics.

---

**Methodological Rigor Supports Reliable Interpretation**
- **PCA** summarizes correlated health indicators (life expectancy, adult mortality, BMI) into a **single composite index**, mitigating multicollinearity.  
- Non-parametric tests, robust regression, and panel methods account for **non-normality, heteroskedasticity, and outliers**, ensuring that conclusions drawn from both **cross-sectional (2015)** and **long-term panel analyses** are statistically valid.

## 4.4-Global Health Overview in 2015

```python
#Regional summary table for 2015 with styling
regional_table_styled = (
    df[df['Year'] == 2015]
    .groupby(['Country', 'Region'], as_index=False)
    .agg(PC1_avg=('PC1_health_index', 'mean'))  
    .groupby('Region')['PC1_avg']
    .agg(
        q1=lambda x: x.quantile(0.25),
        median='median',
        q3=lambda x: x.quantile(0.75),
        min='min',
        max='max'
    )
    .reset_index()
    .style
    .background_gradient(
        cmap='RdYlGn_r',      
        subset=['q1', 'median', 'q3', 'min', 'max']
    )
    .set_caption(
        "Regional Distribution of the Composite Health Index (PC1), 2015<br>"
        "Lower values indicate better population health"
    )
    .format(precision=3)
)

regional_table_styled
```

```python
import plotly.express as px

def plot_health_choropleth(df, country_col='Country', health_col='PC1_health_index', dev_col='Economy_status_Developed'):
    #Copy and map development status
    df = df.copy()
    df['Development_Status'] = df[dev_col].map({0: 'Developing', 1: 'Developed'})

    #Create choropleth
    fig = px.choropleth(
        df,
        locations=country_col,
        locationmode='country names',
        color=health_col,
        hover_name=country_col,
        hover_data=['Development_Status', 'GDP_per_capita', 'Schooling'],
        color_continuous_scale='RdYlGn_r',  
        title='Global Health in 2015 by Country (PC1 Health Index)'
    )

    #Layout adjustments
    fig.update_layout(
        geo=dict(showframe=False, showcoastlines=True, projection_type='natural earth'),
        coloraxis_colorbar=dict(title='PC1 Health Index')
    )

    return fig

#Plot map
fig = plot_health_choropleth(df)
fig.show()
```

The choropleth map above illustrates the **composite health index (PC1)** across countries in 2015, providing a snapshot of overall population health at a global scale. 

Lower values of the PC1 index (shown in green) indicate **better health outcomes**, characterized by higher life expectancy, lower adult mortality, and more favorable nutritional status. Conversely, higher values (shown in red) represent **poorer health outcomes**. 

From the map, we can see clear spatial patterns: many **high-income regions in Europe, North America, and Oceania** display lower PC1 values, signaling better population health. In contrast, parts of **Africa and South Asia** exhibit higher PC1 values, reflecting persistent health challenges.

This visualization highlights that even in 2015, **structural disparities in health remain strongly linked to geography and economic development**, setting the stage for deeper analyses of the factors driving these differences.


## 4.5-Regional Health Disparities in 2015

```python
def plot_regional_health_box_2015(df, health_col='PC1_health_index', region_col='Region'):
    #Average PC1 by country for 2015
    df_country_avg = (
        df[df['Year'] == 2015]
        .groupby(['Country', region_col], as_index=False)
        .agg(PC1_avg=(health_col, 'mean'))
    )

    #Order regions by median health index
    region_order = df_country_avg.groupby(region_col)['PC1_avg'].median().sort_values().index

    #Create boxplot
    fig = px.box(
        df_country_avg,
        x=region_col,
        y='PC1_avg',
        color=region_col,
        category_orders={region_col: region_order},
        title="Regional Disparities in Global Health (Country-Averaged PC1, 2015)",
        labels={'PC1_avg': "PC1 Health Index (avg by country)", region_col: "Region"},
        hover_data=['Country']
    )

    #Layout adjustments
    fig.update_layout(
        showlegend=False,
        xaxis_title="Region",
        yaxis_title="PC1 Health Index",
        boxmode='group'
    )

    return fig

#Usage
fig_box_country_avg_2015 = plot_regional_health_box_2015(df)
fig_box_country_avg_2015.show()
```

```python
#Use the aggregated df_country_avg
df_country_avg_2015 = df[df['Year'] == 2015].groupby(['Country', 'Region'], as_index=False)\
                                          .agg(PC1_avg=('PC1_health_index', 'mean'))

#Identify outliers within each region
outlier_rows_2015 = []
for region, group in df_country_avg_2015.groupby('Region'):
    q1 = group['PC1_avg'].quantile(0.25)
    q3 = group['PC1_avg'].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outliers = group[(group['PC1_avg'] < lower) | (group['PC1_avg'] > upper)]
    outlier_rows_2015.append(outliers)

df_outliers_2015 = pd.concat(outlier_rows_2015).sort_values(['Region', 'PC1_avg'])

#Create a color-coded table
styled_outliers_table_2015 = (
    df_outliers_2015.style
    .background_gradient(
        cmap='RdYlGn_r',   
        subset=['PC1_avg']
    )
    .set_caption(
        "Outlier Countries in Regional Health Index (2015)\n<br>(Lower PC1_avg indicates better population health)"
    )
    .format({'PC1_avg': "{:.3f}"})
)

styled_outliers_table_2015
```

The distribution of the composite health index (PC1_health_index) highlights pronounced regional disparities. Lower values indicate better overall population health, summarizing life expectancy, adult mortality, and BMI.

- **North America** and the **European Union** exhibit the healthiest populations, with medians of -1.66 and -1.37, respectively. Their interquartile ranges are relatively narrow, indicating more homogeneous health outcomes across countries. No outlier countries were identified in these regions. 
- The **Middle East** displays intermediate health outcomes, with a median of -1.30. **Yemen, Rep.** stands out as an outlier with substantially poorer health outcomes relative to the regional norm.  
- **Central America and the Caribbean** have a median of -0.76, with **Haiti** and **Sao Tome and Principe** identified as outliers—Haiti with worse health outcomes and Sao Tome and Principe with better-than-typical outcomes for the region.  
- **Africa** shows the poorest overall health outcomes, with a median of 1.95 and a wide interquartile range. Several countries deviate from the regional median with better health outcomes: **Libya, Tunisia, Egypt, Morocco, Algeria, Seychelles**.  
- Other regions like **Asia**, **South America**, and **Oceania** display broader distributions with no extreme outliers in 2015.

The table of regional outliers complements the boxplots by highlighting the specific countries that deviate from regional norms, emphasizing that structural inequalities persist even within regions.

## 4.7-Summary and Limitations

**Summary**

In 2015, population health disparities are strongly structured by economic development and geography. Developed countries show better overall health outcomes—higher life expectancy, lower mortality, more favorable BMI, and higher schooling while less developed regions systematically lag behind. These patterns are persistent over time (2000–2015) and are robust to non-normality, outliers, and regional heterogeneity.

**Limitations**  
- **Causality**: Analyses are descriptive and correlational; causal inference is limited. High correlation between education and health (ρ ≈ −0.74) prevents clear separation of effects.  
- **Composite index**: PC1_health_index summarizes multiple indicators, potentially masking variation in individual components.  
- **Data quality**: Health and education statistics may be subject to reporting errors or inconsistencies, especially in developing countries.  
- **Temporal coverage**: Panel data only spans 2000–2015; longer-term trends or recent developments are not captured.  
- **Outliers**: Extreme values influence summaries; robust methods mitigate but do not fully remove their impact.  
- **Regional aggregation**: Grouping countries into regions may hide intra-regional disparities.  
- **Model assumptions**: Robust regressions and non-parametric tests reduce sensitivity to non-normality, but residual confounding and multicollinearity remain.  
- **Policy relevance**: Findings describe structural patterns but do not directly identify actionable interventions.
